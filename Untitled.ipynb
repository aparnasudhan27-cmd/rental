```python
messy_df=pd.DataFrame({
    "age" : [24,31,np.nan,29,31,45,27,500],
    "income" : [25000,42000,38000,np.nan,42000,52000,31000,45000],
    "city" : ["kochi","kkd","kochi","thrissur","kkd","kochi",None,"kochi"],
    "purchase_amt" : [1200,2300,1800,1700,2300,2300,900,100000]
});
```

```python
messy_df.describe()
```

```python
messy_df["city"].value_counts()
```

```python
messy_df["city"].value_counts(normalize=True)
```

```python
pip install matplotlib seaborn
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,4))
sns.histplot(messy_df["income"],kde=True)
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=messy_df["income"])
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```python
city_plot=messy_df["city"].fillna("Missing") 
plt.figure(figsize=(8,4))
sns.countplot(x=city_plot,order=city_plot.value_counts().index)
plt.title("City Frequency")
plt.xlabel("City")
plt.ylabel("No of records")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=messy_df,
    x="income",
    y="purchase_amt",
    hue="city",
    s=80
)
plt.title("Income vs purchase amount")
plt.xlabel("income")
plt.ylabel("purchase amount")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=messy_df,
    x="income",
    y="age",
    hue="city",
    s=80
)
plt.title("Income vs age")
plt.xlabel("income")
plt.ylabel("age")
plt.show()
```

```python
from sklearn.impute import SimpleImputer
mean_imputer=SimpleImputer(strategy="mean")
messy_df["income"]=mean_imputer.fit_transform(messy_df[["income"]])
```

```python
display(messy_df)
```

```python
from sklearn.impute import SimpleImputer
median_imputer=SimpleImputer(strategy="median")
messy_df["age"]=median_imputer.fit_transform(messy_df[["age"]])
```

```python
display(messy_df)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=messy_df, x="city" , y="purchase_amt")
plt.title("purchase amount by city")
plt.xlabel("city")
plt.ylabel("purchase_amt")
plt.xticks(rotation=30)
plt.show()
```

```python
pairplot_df=messy_df[["age","income","purchase_amt"]]
sns.pairplot(pairplot_df)
plt.show()
```

```python
numeric_df=messy_df.select_dtypes(include=np.number)
plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(numeric_only=True),annot=True,cmap="coolwarm",center=0)
plt.title("correlation heatmap")
plt.show()
```
