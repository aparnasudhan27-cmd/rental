```python
basic_features=["bathrooms","bedrooms","price","latitude","longitude","feature_count","photo_count","listing_hour","listing_dayofweek"]
x_basic=rental[basic_features].copy()
y=rental["interest_level"].copy()
print("x:",x_basic.shape)
print("y:",y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_basic,y,test_size=.20,random_state=42,stratify=y)
print("training:",x_train.shape)
print("testing:",x_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
numeric_preprocessor=Pipeline([("imputer",SimpleImputer(strategy="median")),("scaler",StandardScaler())])
print("numerical preprocessing pipeline created.")
```

```python
from sklearn.linear_model import LogisticRegression
logistic_basic=Pipeline([("preprocessing",numeric_preprocessor),("model",LogisticRegression(max_iter=1000))
                        ])
logistic_basic.fit(x_train,y_train)
basic_pred=logistic_basic.predict(x_test)
print("Baseline logistic regression trained.")
```

```python
from sklearn.metrics import(accuracy_score,classification_report,confusion_matrix,ConfusionMatrixDisplay,f1_score)
print("accuracy:",round(accuracy_score(y_test,basic_pred),3))
print("macro F1:",round(f1_score(y_test,basic_pred,average="macro"),3))
print("\nclassification report:")
print(classification_report(y_test,basic_pred,zero_division=0))
ConfusionMatrixDisplay.from_predictions(y_test,basic_pred,labels=["low","medium","high"],display_labels=["low","medium","high"])
plt.title("baseline logistic regression")
plt.show()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_demo=TfidfVectorizer(stop_words="english",max_features=20)
text_demo=rental["description"].fillna("").astype(str)
tfidf_matrix_demo=tfidf_demo.fit_transform(text_demo)
print("tf-idf matrix shape:",tfidf_matrix_demo.shape)
print("example vocabulary:")
print(tfidf_demo.get_feature_names_out()[:20])
```

```python
numeric_features=basic_features
text_feature="description"
text_preprocessor=TfidfVectorizer(stop_words="english",max_features=1000,ngram_range=(1,2))
from sklearn.compose import ColumnTransformer
combined_preprocessor=ColumnTransformer(transformers=[("num",numeric_preprocessor,numeric_features),("text",text_preprocessor,text_feature)],remainder="drop")
print("combined numerical+tf-idf preprocessor created")
```

```python
numeric_features = basic_features

text_preprocessor1 = TfidfVectorizer(
    stop_words="english",
    max_features=1000,
    ngram_range=(1,2)
)

text_preprocessor2 = TfidfVectorizer(
    stop_words="english",
    max_features=1000,
    ngram_range=(1,2)
)

from sklearn.compose import ColumnTransformer

combined_preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_preprocessor, numeric_features),
        ("features", text_preprocessor1, "features_text"),
        ("description", text_preprocessor2, "description")
    ],
    remainder="drop"
)

print("combined numerical+tf-idf preprocessor created")
```

```python
x_combined=rental[numeric_features+[text_feature]].copy()
x_train_c,x_test_c,y_train_c,y_test_c=train_test_split(x_combined,y,test_size=.20,random_state=42,stratify=y)
logistic_tfidf=Pipeline([("preprocessing",combined_preprocessor),("model",LogisticRegression(max_iter=1000,C=1.0))])
logistic_tfidf.fit(x_train_c,y_train_c)
tfidf_pred=logistic_tfidf.predict(x_test_c)
print("tf-idf +logistic regression trained.")
print("accuracy:",round(accuracy_score(y_test_c,tfidf_pred),3))
print("macro f1:",round(f1_score(y_test_c,tfidf_pred,average="macro"),3))
print("\nclassification report:")
print(classification_report(y_test_c,tfidf_pred,zero_division=0))
ConfusionMatrixDisplay.from_predictions(y_test_c,tfidf_pred,labels=["low","medium","high"],display_labels=["low","medium","high"])
plt.title("baseline logistic regression")
plt.show()
```

```python
balanced_logistic = Pipeline([
    ("preprocessor", combined_preprocessor),
    ("classifier", LogisticRegression(
        C=1.0,
        class_weight="balanced",
        max_iter=1000
    ))
])

balanced_logistic.fit(x_train_c, y_train_c)

balanced_pred = balanced_logistic.predict(x_test_c)

print("Balanced Logistic Regression")
print("accuracy:", round(accuracy_score(y_test_c, balanced_pred), 4))
print("macro f1:", round(
    f1_score(y_test_c, balanced_pred, average="macro"), 4
))

print(classification_report(
    y_test_c,
    balanced_pred,
    zero_division=0
))
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    balanced_logistic,
    x_train_c,
    y_train_c,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

print("fold macro f1 scores:")
print(cv_scores)

print("\nmean:", round(cv_scores.mean(), 4))
print("standard deviation:", round(cv_scores.std(), 4))
```

```python
rf_preprocessor=ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline([
                ("imputer",SimpleImputer(strategy="median")),
    
            ]),
            numeric_features
        ),
    ],
    remainder="drop"
)
rf_pipeline=Pipeline([
      ("preprocessor",rf_preprocessor),
      ("classifier",RandomForestClassifier(
          n_estimators=200,
          random_state=42,
          n_jobs=-1
      ))
  ])
rf_pipeline.fit(x_train,y_train)
rf_pred=rf_pipeline.predict(x_test)
print("Random forest accuracy",round(accuracy_score(y_test,rf_pred),4))
print("Random forest macro f1",round(f1_score(y_test,rf_pred,average="macro"),4))
```

```python
rf_param_grid={
    "classifier__n_estimators":[100,200],
    "classifier__max_depth":[None,10,20],
    "classifier__min_samples_split":[2,5],
    "classifier__class_weight":[None,"balanced"]
}
rf_grid=GridSearchCV(
    estimator=rf_pipeline,
    param_grid=rf_param_grid,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1,
    verbose=1
)
rf_grid.fit(x_train,y_train)
print("Best Random Forest parameters:")
print(rf_grid.best_params_)
print("\nBest cv macro f1:",round(rf_grid.best_score_,4))
```

```python
rf_tuned_pred = rf_grid.best_estimator_.predict(x_test)

rf_tuned_accuracy = accuracy_score(y_test, rf_tuned_pred)

rf_tuned_macro_f1 = f1_score(y_test, rf_tuned_pred, average = "macro")

print(
    "Tuned RF Accuracy:",
    round(rf_tuned_accuracy, 4)
)

print(
    "Tuned RF Macro F1:",
    round(rf_tuned_macro_f1, 4)
)

print("\nClassification Report:")
print(classification_report(
        y_test,
        rf_tuned_pred
))
```
